```python
class Numbers:
    def __init__(self, pos1, pos2, neg1, neg2):
        self.pos1 = pos1
        self.pos2 = pos2
        self.neg1 = neg1
        self.neg2 = neg2

    def printdata(self):
        print("Positive values:", self.pos1, self.pos2)
        print("Negative values:", self.neg1, self.neg2)

b = Numbers(1, 2, -1, -2)
b.printdata()
```

```python
class Number:
    def _init_(self,n):
        self.number=n
    def check_num(self):
        if self.number>0:
            print("positive")
        else:
            print("negative")


n1=Number(int(input("enter a number")))
n1.check_num()
```
